```python
from pyspark.sql import SparkSession
```

```python
spark = (
    SparkSession.builder
    .appName("TP1P2")
    .getOrCreate()
) 
sc = spark.sparkContext
print("SparkContext initialized with app name:", sc.appName)
```

```python
rdd_people = sc.textFile("/home/jovyan/data_tp1/people.txt")

rdd_people.collect()
```

```python
from pyspark.sql import Row

rdd_people_row = rdd_people.map(
    lambda ligne: Row(
        name=ligne.split(",")[0].strip(),
        age=int(ligne.split(",")[1].strip())
    )
)
```

```python
rdd_people_row.collect()
```

```python
df_people = spark.createDataFrame(rdd_people_row)

df_people.show()
df_people.printSchema()
```

```python
rdd_people = sc.parallelize([
    "Michael, 29",
    "Andy, 30",
    "Justin, 19"
])
```

Instruction	Rôle
map()	Applique une transformation à chaque élément du RDD
lambda l	Représente une ligne du fichier
split(",")	Découpe la ligne au niveau de la virgule
[0]	Récupère le nom
[1]	Récupère l'âge
strip()	Supprime les espaces inutiles
int()	Convertit l'âge en entier
Row()	Crée un objet structuré avec des champs nommés
